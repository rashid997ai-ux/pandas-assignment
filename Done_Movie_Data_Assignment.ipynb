### **Question 15**
'Runtime' column se 'min' text ko hata kar use proper integer (numeric) data type mein convert karein.

```python
df["Runtime"] = df["Runtime"].str.replace(" min", "", regex=False)
df["Runtime"] = pd.to_numeric(df["Runtime"], errors="coerce").astype("Int64")

print(df[["Series_Title", "Runtime"]].head(10))
print("\nRuntime column ka data type:", df["Runtime"].dtype)
```

### **Question 16**

```python
gross_median = df["Gross"].median()

df["Gross"] = df["Gross"].fillna(gross_median)

print("Gross ka median:", gross_median)
print("Gross mein missing values:", df["Gross"].isnull().sum())
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import matplotlib.pyplot as plt

drama_gross = df.loc[
    df["Genre"].str.contains("Drama", case=False, na=False),
    "Gross"
].dropna()

action_gross = df.loc[
    df["Genre"].str.contains("Action", case=False, na=False),
    "Gross"
].dropna()

plt.figure(figsize=(8, 5))

plt.boxplot(
    [drama_gross, action_gross],
    tick_labels=["Drama", "Action"]
)

plt.title("Gross Revenue: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")

plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import numpy as np

top_5 = df.nlargest(5, "IMDB_Rating").copy()


top_5["IMDb_100"] = top_5["IMDB_Rating"] * 10

x = np.arange(len(top_5))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    x - width / 2,
    top_5["Meta_score"],
    width,
    label="Meta Score"
)

plt.bar(
    x + width / 2,
    top_5["IMDb_100"],
    width,
    label="IMDb Rating (x10)"
)

plt.title("Top 5 Highest-Rated Movies: Meta Score vs IMDb Rating")
plt.xlabel("Movie Name")
plt.ylabel("Score out of 100")

plt.xticks(x, top_5["Series_Title"], rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```
